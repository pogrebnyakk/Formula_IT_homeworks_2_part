Series

```python
data = pd.Series([1,2,3,4])
print(data)
print(data.values) #сами данные
print(data.index) # индексы по которым происходит обращение
print(type(data.values))
print(type(data.index))
```

RangeIndex является итератором. Массив значений values является типом numpy array, index - pandas index

```python
print(data[0])
print(data[1:3])
```

Можно самостоятельно определить индекс у массива. При этом тип индекса становится pandas indexes. В качестве индексов может быть любой объект, в качестве срезов при этом можно также использовать эти объекты

Ведет себя как одномерный массив. Срезы можно задавать в явном и неявном виде(порядковый номер)

```python
data = pd.Series([1,2,3,4], index = ['a', 'a', 'c', 'd'])
print(data)
print(data['a'])
print(data['c':'d'])
print(data[2:3])
print(type(data.index))

```

Можно фильтровать данные по различным условиям. Также возможные некоторые функции словаря.

```python
data = pd.Series([1,2,3,4], index = ['a', 'a', 'c', 'd'])
print('a' in data.index)
print('a' in data)
print(data.keys())
print(list(data.items()))
print(data[data > 2])
```

Если в качестве index используются int'ы, необходимо использовать атрибуты индексаторы, чтобы показать системе, как разрешить обращение (к чему обращаться)

```python
data = pd.Series([213213,323,312,544], index = [1, 2, 3, 4])
print(data)
print(data.loc[1])
print(data.iloc[1])
```

Для создания Series можно использовать:  
1) Списки или массивы NumPy
2) Скалярные значения (будет повторять в заданном размере массива)
3) Словари

```python
population = {
    'city1': 1000,
    'city2': 1001,
    'city3': 1002,
    'city4': 1003,
    'city5': 1004,
}
populate = pd.Series(population)
print(populate['city1'])
print(populate['city3':'city4'])
```

#### DataFrame

Является двумерным массивом с явно определенными индексами. Последовательность "согласованных" объектов Series.

Можно создавать через объекты:
1) Series
2) Списки словарей
3) Словари объектов Series
4) Двумерный массив NumPy
5) Структурированный массив NumPy

```python
population = {
    'city1': 1000,
    'city2': 1001,
    'city3': 1002,
    'city4': 1003,
    'city5': 1004,
}

area = {
    'city1': 45678,
    'city2': 2345,
    'city3': 123456,
    'city4': 7890,
    'city5': 4567,
}

populate = pd.Series(population)
area = pd.Series(area)
states = pd.DataFrame({'population': populate, 'area': area})
print(states)
print(states.values)
print(states.index)
print(states.columns)
print(type(states.values), type(states.index), type(states.columns))
```

```python
print(states['area'])
```

Можно обращаться к столбцу через точку. Ведет себя как словарь или двумерный NumPy массив.

```python
population = {
    'city1': 1000,
    'city2': 1001,
    'city3': 1002,
    'city4': 1003,
    'city5': 1004,
}

area = {
    'city1': 45678,
    'city2': 2345,
    'city3': 123456,
    'city4': 7890,
    'city5': 4567,
}

data = pd.DataFrame({'area': area, 'population': population})
print(data)
print(data['area'] is data.area)

data['new'] = data['area']
data['new1'] = data['area'] / data['population']
print(data)
```

#### Объединение датафреймов

```python
import numpy as np
rng = np.random.default_rng()
dfA = pd.DataFrame(rng.integers(0,10,(2,2)), columns = ['a', 'b'])
dfB = pd.DataFrame(rng.integers(0,10,(3,3)), columns = ['a', 'b', 'c'])
dfC = dfA + dfB

print(dfC)
```

Создаем матрицу 3х4. В numpy и pandas используется $транслирование$. Например, вычтем из матрицы 3х4 строку 0, тогда от каждого значения соответственно будет вычтено это значение.

```python
rng = np.random.default_rng(1)
A = rng.integers(0, 10, (3,4))
print(A)
print(A[0])
print(A - A[0])

df = pd.DataFrame(A, index = ['a', 'b', 'c'])
print(df - df.iloc[0])
print(df - df.iloc[0, ::2]) # нечетные не участвуют в вычислениях
```

### Объект Index
Это способ организации ссылки на данные объектов Series и DataFrame. Явялется неизменяемым и упорядочей(на самом деле каждый объект идет по порядку 1,2,3), также является мультимножеством (могут быть повторяющиеся значения)

Также индекс следует соглашениям объекта set (можно пересекать, объединять и т.д.)

```python
ind = pd.Index([2,3,4,5,1,2])
print(ind[0])
print(ind[::2])
indA = pd.Index([1,2,3,4,5])
indB = pd.Index([7,3,4,5])
print(indA.intersection(indB))
```

NaN - not a number

NA - значения: NaN, null

Нужно уметь работать с отсутствующими значениями

В pandas два способа хранения отсутствующих значений:
1) Использование индикаторов - NaN, None 
2) null

None - Объект (накладные расходы). Не работает с агрегирующими операторами. Если присутствуют различные отсутствующие значения, можно их привести в NA с помощью dtype='int32'. 

Также можно использовать идентификацию таких элмеентов с помощью .notnull или .isnull. Избавиться от них можно с помощью .dropna(), также можно указать размерность, с которой осуществляем выбрасывание. С помощью параметра how(принимает параметры: all - строка и столбец отбрасывается, если отсутствуют все значения; any - хотя бы одно значение; thresh - порог = х, строка остается, если присутствует минимум х непустых значений)

```python
x = pd.Series([1,2,3,np.nan,None,pd.NA])
x
print(x.isnull())
print(x.notnull())
print(x.dropna())

df = pd.DataFrame([[1,2,3,np.nan,None,pd.NA], [1,2,3,4,5], [np.nan,1,2,3]])
print(df.dropna(axis=1))
```
